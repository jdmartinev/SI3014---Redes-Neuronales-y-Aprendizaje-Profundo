```python
# 1. Define the transformation: Convert the image to a PyTorch tensor
#    This transform also scales pixel values from [0, 255] to [0.0, 1.0]
transform = transforms.Compose([transforms.ToTensor()])
# 2. Load the image using PIL
img_path = 'perro.png' # Replace with your image path
image_pil = Image.open(img_path).convert('RGB') # Open image and ensure RGB mode
# 3. Apply the transform to convert to a tensor
image_tensor = transform(image_pil)

# The tensor will have the shape [C, H, W] (Channels, Height, Width)
print(image_tensor.shape)
```
